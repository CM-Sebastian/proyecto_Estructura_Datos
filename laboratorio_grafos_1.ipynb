````python
#Paso 2
class Grafo :
    def __init__ ( self ) :
        self . vertices = []
        self . aristas = {}
    #Paso 3
    def agregar_vertice ( self , vertice ) :
        if vertice not in self . vertices :
            self . vertices . append ( vertice )
            print (" Vertice agregado :", vertice )
        else :
           print ("El vertice ya existe :", vertice )

    #Paso 4
    def agregar_arista ( self , nombre , origen , destino ):
        if nombre in self . aristas :
            print ("Ya existe una arista llamada ", nombre )
            return

        if origen not in self . vertices :
            print ("No existe el vertice :", origen )
            return

        if destino not in self . vertices :
            print ("No existe el vertice :", destino )
            return

        self . aristas [ nombre ] = ( origen , destino )
        print (
        " Arista agregada :",
        nombre ,
        "=",
        origen ,
        " --",
        destino
        )

    #Paso 6
    def mostrar_grafo ( self ):
        print ("\n- -- GRAFO ---")

        print (" Vertices :")
        for vertice in self . vertices :
            print ("-", vertice )

        print ("\nAristas :")
        for nombre , extremos in self . aristas . items () :
            origen , destino = extremos
            print ( nombre , ":", origen , " --", destino )

    #Paso 7
    def es_bucle ( self , nombre_arista ) :
        if nombre_arista not in self . aristas :
            return False

        origen , destino = self . aristas [ nombre_arista ]
        return origen == destino

    #Paso 8
    def son_paralelas ( self , arista1 , arista2 ) :
        if arista1 not in self . aristas :
            return False

        if arista2 not in self . aristas :
            return False

        origen1 , destino1 = self . aristas [ arista1 ]
        origen2 , destino2 = self . aristas [ arista2 ]
        mismos_extremos = (
        origen1 == origen2 and destino1 == destino2
        )

        extremos_invertidos = (
        origen1 == destino2 and destino1 == origen2
        )

        return mismos_extremos or extremos_invertidos

    #Paso 9
    def obtener_aristas_paralelas ( self ):
        paralelas = []
        nombres = list ( self . aristas . keys () )

        for i in range ( len ( nombres ) ):
            for j in range (i + 1 , len ( nombres ) ):
                arista1 = nombres [i]
                arista2 = nombres [j]

                if self . son_paralelas ( arista1 , arista2 ):
                    paralelas . append (( arista1 , arista2 ) )

        return paralelas

    #Paso 10
    def grado ( self , vertice ):
        if vertice not in self . vertices :
            return None

        contador = 0

        for origen , destino in self . aristas . values () :

            if origen == vertice and destino == vertice :
                contador += 2
            elif origen == vertice or destino == vertice :
                contador += 1

        return contador

    #Paso 11
    def es_aislado ( self , vertice ):
        grado_vertice = self . grado ( vertice )

        if grado_vertice is None :
            return False

        return grado_vertice == 0

    def obtener_vertices_aislados ( self ):
        aislados = []

        for vertice in self . vertices :
            if self . es_aislado ( vertice ) :
                aislados . append ( vertice )

        return aislados

    #Paso 12
    def grado_total ( self ):
        total = 0

        for vertice in self . vertices :
            total += self . grado ( vertice )

        return total

    #Paso 13
    def mostrar_grados ( self ) :
        print ("\n- -- GRADOS DE LOS VERTICES ---")

        for vertice in self . vertices :
            print (
            vertice ,
            ":",
            self . grado ( vertice )
            )

        print (
        " Grado total :",
        self . grado_total ()
    )

    #Paso 14
    def es_simple ( self ) :
        for nombre in self . aristas :
            if self . es_bucle ( nombre ):
                return False

        if len( self . obtener_aristas_paralelas () ) > 0:
            return False

        return True

    #PAso 15
    def matriz_adyacencia ( self ) :
        cantidad = len ( self . vertices )

        matriz = []

        for i in range ( cantidad ):
            fila = [0] * cantidad
            matriz . append ( fila )

        for origen , destino in self . aristas . values () :
            i = self . vertices . index ( origen )
            j = self . vertices . index ( destino )

            if i == j :
                matriz [i ][ j ] += 2
            else :
                matriz [i ][ j ] += 1
                matriz [j ][ i ] += 1

        return matriz

    #Paso 16
    def mostrar_matriz_adyacencia ( self ):
        matriz = self . matriz_adyacencia ()

        print ("\n- -- MATRIZ DE ADYACENCIA ---")

        print (" " * 14 , end ="")

        for vertice in self . vertices :
            print (f"{ vertice [:8]: >10} ", end ="")

        print ()

        for i in range ( len ( self . vertices )):
            print (f"{ self . vertices [i ][:12]: >12} ", end ="    ")

            for valor in matriz [i ]:
                print ( f"{ valor : >10}", end =" ")

            print ()


    #Paso 17
    def matriz_incidencia ( self ) :
        cantidad_vertices = len ( self . vertices )
        cantidad_aristas = len ( self . aristas )

        matriz = []

        for i in range ( cantidad_vertices ):
            fila = [0] * cantidad_aristas
            matriz . append ( fila )

        nombres_aristas = list ( self . aristas . keys () )

        for j in range ( cantidad_aristas ) :
            nombre = nombres_aristas [j ]
            origen , destino = self . aristas [ nombre ]

            i_origen = self . vertices . index ( origen )
            i_destino = self . vertices . index ( destino )

            if i_origen == i_destino :
                matriz [ i_origen ][ j ] = 2
            else :
                matriz [ i_origen ][ j ] = 1
                matriz [ i_destino ][ j] = 1

        return matriz

    #Paso 18
    def mostrar_matriz_incidencia ( self ):
        matriz = self . matriz_incidencia ()
        nombres_aristas = list ( self . aristas . keys () )

        print ("\n- -- MATRIZ DE INCIDENCIA ---")

        print (" " * 14 , end ="")

        for nombre in nombres_aristas :
            print (f"{ nombre : >6}", end ="")

        print ()

        for i in range ( len ( self . vertices )):
            print (f"{ self . vertices [i ][:12]: >12} ", end =" ")

            for valor in matriz [i ]:
                print ( f"{ valor : >6}", end ="")

            print ()


    #Paso 19
    def obtener_vecinos ( self , vertice ):
        if vertice not in self . vertices :
            return []

        vecinos = []

        for origen , destino in self . aristas . values () :
            if origen == vertice and destino not in vecinos :
                vecinos . append ( destino )

            if destino == vertice and origen not in vecinos :
                vecinos . append ( origen )

        return vecinos

    #Paso 20
    def lista_adyacencia ( self ):
        lista = {}

        for vertice in self . vertices :
            lista [ vertice ] = []

        for origen , destino in self . aristas . values () :
            lista [ origen ]. append ( destino )

            if origen != destino :
                lista [ destino ]. append ( origen )

        return lista


    def mostrar_lista_adyacencia ( self ) :
        lista = self . lista_adyacencia ()

        print ("\n- -- LISTA DE ADYACENCIA ---")

        for vertice , vecinos in lista . items () :
            print ( vertice , " ->", vecinos )


    #Paso 21
    def resumen ( self ) :
        print ("\n ========== RESUMEN DEL GRAFO ========== ")

        print (" Cantidad de vertices :", len ( self . vertices ))
        print (" Cantidad de aristas :", len ( self . aristas ) )
        print (" Grado total :", self . grado_total () )
        print ("Es simple :", self . es_simple () )

        print (
        " Vertices aislados :",
        self . obtener_vertices_aislados ()
        )

        print (
        " Aristas paralelas :",
        self . obtener_aristas_paralelas ()
        )

        bucles = []

        for nombre in self . aristas :
            if self . es_bucle ( nombre ):
                bucles . append ( nombre )

        print (" Bucles :", bucles )

#Paso 3
#grafo = Grafo ()
"""
grafo . agregar_vertice (" Bodega ")
grafo . agregar_vertice (" Norte ")
grafo . agregar_vertice (" Centro ")
grafo . agregar_vertice (" Bodega ")

print ( grafo . vertices )"""

#Paso 5
"""
grafo . agregar_vertice (" Bodega ")
grafo . agregar_vertice (" Norte ")
grafo . agregar_vertice (" Centro ")
grafo . agregar_vertice ("Sur")
grafo . agregar_vertice (" Aeropuerto ")

grafo . agregar_arista ("e1", " Bodega ", " Norte ")
grafo . agregar_arista ("e2", " Bodega ", " Centro ")
grafo . agregar_arista ("e3", " Norte ", " Centro ")
grafo . agregar_arista ("e4", " Centro ", " Sur ")
"""
"""print ("\nVertices :")
print ( grafo . vertices )

print ("\nAristas :")
print ( grafo . aristas )"""

#Paso 6
#grafo . mostrar_grafo ()

#Paso 7
#grafo . agregar_arista ("e5", "Sur", "Sur")
"""
print (
"e4 es bucle :", grafo . es_bucle ("e4"))
print ("e5 es bucle :", grafo . es_bucle ("e5"))"""

#Paso 8
#grafo . agregar_arista ("e6", " Norte ", " Bodega ")
"""
print (
"e1 y e6 son paralelas :",
grafo . son_paralelas ("e1", "e6")
)

print (
"e2 y e3 son paralelas :",
grafo . son_paralelas ("e2", "e3")
)"""

#Paso 9
"""
pares = grafo . obtener_aristas_paralelas ()

print (" Aristas paralelas encontradas :")
for par in pares :
    print ( par )
"""
#Paso 10
"""
grafo_simple = Grafo ()

grafo_simple . agregar_vertice ("Bodega")
grafo_simple . agregar_vertice ("Norte")
grafo_simple . agregar_vertice ("Centro")
grafo_simple . agregar_vertice ("Sur")
grafo_simple . agregar_vertice ("Aeropuerto")

grafo_simple . agregar_arista (
"e1", "Bodega", "Norte"
)
grafo_simple . agregar_arista (
"e2", "Bodega", "Centro"
 )
grafo_simple . agregar_arista (
"e3", "Norte", "Centro"
)
grafo_simple . agregar_arista (
"e4", "Centro", "Sur"
)

print (
" Grado de Bodega :",
grafo_simple . grado ("Bodega")
)

print (
" Grado de Norte :",
grafo_simple . grado ("Norte")
)

print (

" Grado de Centro :",
grafo_simple . grado ("Centro")
)
print (
" Grado de Sur :",
grafo_simple . grado ("Sur")
)

print (
" Grado de Aeropuerto :",
grafo_simple . grado ("Aeropuerto")

)
"""
#Paso 11
"""
print (
" Bodega esta aislado :",
grafo_simple . es_aislado ("Bodega")
)

print (
" Aeropuerto esta aislado :",
grafo_simple . es_aislado ("Aeropuerto")
)
"""


#Paso 12
"""print (
" Grado total del grafo :",
grafo_simple . grado_total ()
)"""


#Paso 13
#grafo_simple . mostrar_grados ()

#Paso 14

"""print (
" grafo_simple es simple :",
grafo_simple . es_simple ()
)
print (
" grafo es simple :",
grafo . es_simple ()
)"""

#Paso 16

#grafo_simple . mostrar_matriz_adyacencia ()

#Paso 18
#grafo_simple . mostrar_matriz_incidencia ()

#Paso 19
"""print (
" Vecinos de Centro :",
grafo_simple . obtener_vecinos ("Centro")
)

print (
" Vecinos de Aeropuerto :",
grafo_simple . obtener_vecinos ("Aeropuerto")
)"""

#Paso 20

#grafo_simple . mostrar_lista_adyacencia ()


#Paso 21

#grafo_simple . resumen ()

#Prueba de grafo completo

grafo = Grafo ()

grafo . agregar_vertice (" Bodega ")
grafo . agregar_vertice (" Norte ")
grafo . agregar_vertice (" Centro ")
grafo . agregar_vertice (" Sur ")
grafo . agregar_vertice (" Aeropuerto ")

grafo . agregar_arista (
"e1",
" Bodega ",
 " Norte "
)
grafo . agregar_arista (
"e2",
" Bodega ",
" Centro "
)

grafo . agregar_arista (
"e3",
" Norte ",
" Centro "
)


grafo . agregar_arista (
"e4",
" Centro ",
" Sur "
)
#Modificacioes

# Modificacion 1
grafo . agregar_arista (
"e5",
" Sur ",
" Aeropuerto "
)

# Modificación 2
grafo . agregar_arista (
"e6",
" Norte ",
" Bodega "
)

#Modificacion 3
grafo . agregar_arista (
"e7",
" Centro ",
" Centro "
)


grafo . mostrar_grafo ()
grafo . mostrar_grados ()
grafo . mostrar_lista_adyacencia ()
grafo . mostrar_matriz_adyacencia ()
grafo . mostrar_matriz_incidencia ()
grafo . resumen ()
````

#Preguntas de analisis

Responda con base en el código desarrollado:


###1. ¿Qué estructura de Python representa el conjunto de vértices?
Una lista (`self.vertices`).
###2. ¿Qué estructura representa las aristas?
Un diccionario (`self.aristas`) donde la clave es el nombre de la arista y el valor es una tupla `(origen, destino)`.
###3. ¿Cómo se representa la función punto extremo–arista?
Implicitamente, mediante los valores del diccionario `self.aristas`, donde cada arista (clave) se asocia a un par de vértices (tupla de valor).
###4. ¿Por qué cada arista necesita un nombre único?
Para poder identificar y manipular aristas individualmente, especialmente para aristas paralelas o bucles.

###5. ¿Por qué no sería suficiente almacenar solamente pares de vértices en un conjunto?
Porque un conjunto de pares no permitiría distinguir aristas paralelas (mismos vértices, pero aristas distintas) ni identificar aristas individualmente por un nombre.
###6. ¿Cómo permite la implementación conservar aristas paralelas?
Cada arista se almacena con un `nombre` único en el diccionario `self.aristas`, lo que permite que múltiples aristas tengan los mismos vértices de origen y destino.
###7. ¿Cómo se reconoce un bucle?
En el método `es_bucle`, se reconoce si el vértice de origen y destino de una arista son el mismo.
###8. ¿Cuánto aporta un bucle al grado de un vértice?
Aporta 2 unidades al grado de un vértice.
###9. ¿Cómo se identifica un vértice aislado?
Un vértice es aislado si su grado es 0, lo cual se verifica en `es_aislado`.
###10. ¿Qué condiciones debe cumplir un grafo simple?
No debe tener bucles ni aristas paralelas.
###11. ¿Por qué la matriz de adyacencia de un grafo no dirigido es simétrica?
Porque si existe una arista entre A y B, la conexión es mutua (A a B y B a A), resultando en `matriz[A][B] == matriz[B][A]`.
###12. ¿Qué representa una fila de la matriz de adyacencia?
Una fila representa las conexiones (adyacencia) de un vértice específico con todos los demás vértices del grafo.
###13. ¿Qué representa una columna de la matriz de incidencia?
Una columna representa las incidencias de una arista específica con los vértices del grafo.
###14. ¿Cómo se representa un bucle en la matriz de incidencia?
Se representa con un 2 en la intersección de la fila del vértice y la columna del bucle.
###15. ¿Qué diferencia existe entre una matriz de adyacencia y una lista de adyacencia?
La matriz de adyacencia es una tabla cuadrada que muestra conexiones directas (1s o 2s para bucles/paralelas) entre cada par de vértices, mientras que la lista de adyacencia es un diccionario donde cada vértice tiene una lista de sus vecinos directos. La lista es más eficiente para grafos dispersos.
###16. ¿Cómo se puede calcular el grado de un vértice utilizando la matriz de adyacencia?
Sumando todos los valores de la fila (o columna) correspondiente al vértice. Los bucles se cuentan dos veces.
###17. ¿Qué relación existe entre el grado total y el número de aristas?
El grado total es el doble del número de aristas en un grafo no dirigido (considerando que los bucles suman 2 al grado).
###18. ¿Qué podría representar un vértice aislado dentro del proyecto?
Un punto de interés (ej. una tienda, un centro de distribución) que no tiene rutas conectadas en el modelo actual.
###19. ¿Qué podría representar un vértice de grado alto dentro del proyecto?
Un centro de conexión importante, un nodo con muchas rutas o interacciones (ej. un centro de distribución principal o una estación central).
###20. ¿Por qué los grafos son estructuras no lineales?
Porque no tienen un inicio y un fin definidos, y los nodos pueden estar interconectados de múltiples formas, formando estructuras complejas sin una secuencia única.
