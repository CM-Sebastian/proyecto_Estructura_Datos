#Preguntas de analisis
1. ¿Qué función conserva la lista enlazada dentro del sistema?

>La `ListaEnlazada` se encarga de almacenar y gestionar la colección de estudiantes registrados en el sistema. Permite operaciones como insertar, mostrar, buscar y eliminar estudiantes de manera eficiente, manteniendo un orden secuencial.

2. ¿Por qué el historial se implementó mediante una pila?

>El historial se implementó como una `Pila` (Stack) porque una pila sigue el principio LIFO (Last-In, First-Out). Esto es ideal para un historial de acciones, ya que la acción más reciente es la primera que se consulta (`peek`) o se desea deshacer (`pop`).

3. ¿Por qué las solicitudes se implementaron mediante una cola?

>Las solicitudes se implementaron como una `Cola` (Queue) porque una cola sigue el principio FIFO (First-In, First-Out). Esto es apropiado para gestionar solicitudes de atención, donde la primera solicitud en llegar es la primera que debe ser atendida.

4. ¿Cuál es la diferencia entre pop() y dequeue()?

> `pop()` es un método de la `Pila` que elimina y devuelve el elemento superior (el último en ser añadido). `dequeue()` es un método de la `Cola` que elimina y devuelve el elemento frontal (el primero en ser añadido).

5. ¿Qué elemento consulta peek() en una pila?

>En una `Pila`, `peek()` consulta y devuelve el elemento superior, es decir, el último elemento que se añadió, sin eliminarlo.

6. ¿Qué elemento consulta peek() en una cola?

>En una `Cola`, `peek()` consulta y devuelve el elemento frontal, es decir, el primer elemento que se añadió, sin eliminarlo.

7. ¿Por qué se debe verificar is_empty() antes de eliminar elementos?

>Se debe verificar `is_empty()` antes de eliminar elementos para evitar errores (como acceder a un índice fuera de rango) o comportamientos inesperados cuando se intenta eliminar de una estructura de datos vacía. Garantiza la robustez del código.

8. ¿Qué evidencia de las pruebas demuestra el principio LIFO?

>El principio LIFO se demuestra en el historial, donde al mostrar las acciones, la última acción registrada aparece en la cima (la primera en ser listada). Por ejemplo, en la salida de `historial.mostrar()`, si se registran "A", "B", "C" en ese orden, el `mostrar()` listará "C", "B", "A".

9. ¿Qué evidencia de las pruebas demuestra el principio FIFO?

>El principio FIFO se demuestra en las `solicitudes` de la `Cola`. Al añadir solicitudes y luego atenderlas con `dequeue()`, la primera solicitud que se añadió es la primera en ser atendida, como se ve en los pasos de atención de solicitudes.

10. ¿Por qué se verifica que un estudiante exista antes de registrar una solicitud?

>Se verifica que un estudiante exista antes de registrar una solicitud (`estudiantes.buscar(nombre)`) para asegurar que solo se puedan generar solicitudes para estudiantes válidamente registrados en el sistema, evitando solicitudes para entidades inexistentes y manteniendo la integridad de los datos.

Segundo, llama a `historial.push("Registrar estudiante: " + nombre)`, que añade un registro de esta acción a la `Pila` `historial`.

12. ¿Qué ventaja tiene separar el sistema en las clases ListaEnlazada, Pila y Cola?

>La ventaja de separar el sistema en clases como `ListaEnlazada`, `Pila` y `Cola` es la **modularidad** y **reusabilidad**. Cada clase encapsula la lógica de su estructura de datos, haciendo el código más organizado, fácil de entender, mantener y depurar. Además, estas estructuras pueden ser reutilizadas en otras partes del sistema o en diferentes proyectos sin afectar la implementación de las demás.

13. ¿Qué problemas podrían aparecer si todas las operaciones se almacenaran únicamente en una lista de Python?

>Si todas las operaciones se almacenaran en una única lista de Python, los problemas incluirían:
>1.  **Eficiencia**: Las operaciones de `pop(0)` (simulando `dequeue`) o `insert(0)` (simulando `insertar_inicio`) en listas de Python son O(n), lo que sería ineficiente para grandes volúmenes de datos. Las pilas y colas suelen ser O(1) para sus operaciones principales.
>2.  **Claridad**: El código sería menos legible y más propenso a errores, ya que las intenciones de LIFO o FIFO no serían explícitas y se dependería de convenciones de indexación.
>3.  **Mantenimiento**: Sería más difícil modificar o depurar el código debido a la falta de encapsulamiento y la mezcla de lógicas.
>4.  **Cumplimiento de principios**: Sería más fácil violar los principios LIFO/FIFO accidentalmente, ya que la lista de Python ofrece acceso aleatorio.
